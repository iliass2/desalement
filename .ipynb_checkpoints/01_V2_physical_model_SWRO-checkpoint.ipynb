# Notebook 01_V2 — Modèle physique amélioré SWRO & génération du dataset

**PFE — Master Science des Données**

### Objectif

Construire une version améliorée du simulateur physique simplifié d'un procédé de dessalement d'eau de mer par osmose inverse (SWRO).

Cette V2 introduit :

- la pression osmotique simplifiée ;
- une pression nette disponible pour la membrane ;
- un flux de perméat dépendant de la pression, de la température et de l'état de membrane ;
- un modèle énergétique de pompe ;
- une récupération d'énergie sur le concentrat ;
- une vérification des contraintes physiques ;
- un bruit de mesure ;
- une analyse des relations entre variables et SEC.

> **Important :** les données générées sont synthétiques. Les équations et coefficients simplifiés sont des hypothèses de simulation et ne doivent pas être présentés comme des mesures d'une station réelle.

```python
# ============================================================
# 0. IMPORTS ET CONFIGURATION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

np.random.seed(42)

N_SAMPLES = 10_000

OUTPUT_DIR = Path("data")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "dataset_physical_v2.csv"

# Constantes
BAR_TO_PA = 100_000
M3H_TO_M3S = 1 / 3600

RHO = 1025          # kg/m3, densité approximative eau de mer
G = 9.81            # m/s2

# Pression osmotique de référence.
# Pour une simulation simplifiée, on utilise environ 0.75 bar
# par g/L de salinité autour de l'eau de mer.
OSMOTIC_COEFF_BAR_PER_G_L = 0.75

print("Configuration chargée.")
print(f"Nombre d'observations : {N_SAMPLES}")
print(f"Fichier : {OUTPUT_FILE}")
```

## 1. Plages de fonctionnement

Les plages sont des **hypothèses de simulation** destinées à représenter un fonctionnement plausible d'un système SWRO.

Elles ne constituent pas des données de la station.

```python
RANGES = {
    "pressure_bar": (55, 70),
    "feed_flow_m3_h": (200, 1000),
    "salinity_g_L": (34, 42),
    "temperature_C": (15, 30),
    "recovery_rate": (0.30, 0.45),
    "pump_efficiency": (0.75, 0.90),
    "erd_efficiency": (0.80, 0.96),
    "membrane_condition": (0.75, 1.00),
}

for variable, (minimum, maximum) in RANGES.items():
    print(f"{variable:25s}: {minimum} → {maximum}")
```

## 2. Génération des conditions de fonctionnement

```python
pressure_bar = np.random.uniform(*RANGES["pressure_bar"], N_SAMPLES)
feed_flow_m3_h = np.random.uniform(*RANGES["feed_flow_m3_h"], N_SAMPLES)
salinity_g_L = np.random.uniform(*RANGES["salinity_g_L"], N_SAMPLES)
temperature_C = np.random.uniform(*RANGES["temperature_C"], N_SAMPLES)
recovery_rate = np.random.uniform(*RANGES["recovery_rate"], N_SAMPLES)
pump_efficiency = np.random.uniform(*RANGES["pump_efficiency"], N_SAMPLES)
erd_efficiency = np.random.uniform(*RANGES["erd_efficiency"], N_SAMPLES)
membrane_condition = np.random.uniform(*RANGES["membrane_condition"], N_SAMPLES)

print("Variables générées.")
```

## 3. Pression osmotique simplifiée

On utilise une approximation linéaire :

\[
\pi \approx k_\pi S
\]

où :

- \(\pi\) est la pression osmotique en bar ;
- \(S\) est la salinité en g/L ;
- \(k_\pi\) est un coefficient simplifié.

La pression osmotique du flux d'alimentation est ensuite comparée à la pression appliquée par la pompe.

```python
osmotic_pressure_feed_bar = (
    OSMOTIC_COEFF_BAR_PER_G_L * salinity_g_L
)

# Approximation de la salinité moyenne du côté concentrat.
# Plus le taux de récupération est élevé, plus la concentration augmente.
concentrate_salinity_g_L = (
    salinity_g_L / (1 - recovery_rate)
)

osmotic_pressure_concentrate_bar = (
    OSMOTIC_COEFF_BAR_PER_G_L * concentrate_salinity_g_L
)

print(f"Pression osmotique alimentation moyenne : "
      f"{osmotic_pressure_feed_bar.mean():.2f} bar")

print(f"Pression osmotique concentrat moyenne : "
      f"{osmotic_pressure_concentrate_bar.mean():.2f} bar")
```

## 4. Pression osmotique moyenne de la membrane

Pour simplifier le bilan à travers la membrane, on utilise la moyenne des pressions osmotiques de l'alimentation et du concentrat.

\[
\pi_{avg}=\frac{\pi_{feed}+\pi_{conc}}{2}
\]

Puis :

\[
\Delta P_{net}=P_{RO}-\pi_{avg}
\]

La pression nette disponible doit rester positive.

```python
osmotic_pressure_avg_bar = (
    osmotic_pressure_feed_bar
    + osmotic_pressure_concentrate_bar
) / 2

net_membrane_pressure_bar = (
    pressure_bar - osmotic_pressure_avg_bar
)

# Sécurité numérique : pression nette minimale positive.
net_membrane_pressure_bar = np.maximum(
    net_membrane_pressure_bar,
    1.0
)

print(f"Pression nette moyenne : "
      f"{net_membrane_pressure_bar.mean():.2f} bar")
```

## 5. Effet de la température

La température influence la perméabilité de la membrane.

Pour la simulation, on définit un facteur relatif autour de 25 °C.

```python
temperature_reference_C = 25.0

temperature_factor = (
    1 + 0.025 * (temperature_C - temperature_reference_C)
)

temperature_factor = np.clip(
    temperature_factor,
    0.75,
    1.15
)

print(
    f"Facteur température : "
    f"{temperature_factor.min():.3f} → "
    f"{temperature_factor.max():.3f}"
)
```

## 6. Flux de perméat simplifié

On utilise une relation de type :

\[
J = A \cdot \Delta P_{net} \cdot f_T \cdot f_M
\]

où :

- \(J\) est le flux de perméat ;
- \(A\) est un coefficient de perméabilité simplifié ;
- \(f_T\) représente la température ;
- \(f_M\) représente l'état de la membrane.

Le débit de perméat est ensuite limité par le taux de récupération demandé.

```python
# Flux de base choisi pour la simulation.
# Unités simplifiées : L/m2/h par bar de pression nette.
A_base = 0.90

# Dégradation : une membrane à 75 % de condition
# présente une perméabilité plus faible.
membrane_factor = (
    0.75 + 0.25 * membrane_condition
)

flux_permeate_L_m2_h = (
    A_base
    * net_membrane_pressure_bar
    * temperature_factor
    * membrane_factor
)

# Surface membranaire virtuelle.
# Elle varie avec le débit d'alimentation afin de garder
# une échelle cohérente dans la simulation.
membrane_area_m2 = (
    0.90 * feed_flow_m3_h
)

permeate_flow_from_membrane_m3_h = (
    flux_permeate_L_m2_h
    * membrane_area_m2
    / 1000
)

# Débit demandé par le taux de récupération.
permeate_flow_target_m3_h = (
    feed_flow_m3_h * recovery_rate
)

# Le débit réellement produit est le minimum des deux.
permeate_flow_m3_h = np.minimum(
    permeate_flow_from_membrane_m3_h,
    permeate_flow_target_m3_h
)

# On évite des débits quasi nuls.
permeate_flow_m3_h = np.maximum(
    permeate_flow_m3_h,
    1.0
)

effective_recovery_rate = (
    permeate_flow_m3_h / feed_flow_m3_h
)

print(f"Débit perméat moyen : {permeate_flow_m3_h.mean():.2f} m³/h")
print(f"Taux de récupération effectif moyen : "
      f"{effective_recovery_rate.mean()*100:.2f} %")
```

## 7. Débit de concentrat

\[
Q_{conc}=Q_{feed}-Q_{perm}
\]

```python
concentrate_flow_m3_h = (
    feed_flow_m3_h - permeate_flow_m3_h
)

print(f"Débit concentrat moyen : "
      f"{concentrate_flow_m3_h.mean():.2f} m³/h")
```

## 8. Puissance hydraulique de la pompe

\[
P_{hyd}=\Delta P\times Q
\]

La pression appliquée par la pompe est convertie en Pa et le débit d'alimentation en m³/s.

```python
pressure_pa = pressure_bar * BAR_TO_PA
feed_flow_m3_s = feed_flow_m3_h * M3H_TO_M3S

hydraulic_power_W = (
    pressure_pa * feed_flow_m3_s
)

hydraulic_power_kW = (
    hydraulic_power_W / 1000
)

pump_power_kW = (
    hydraulic_power_kW / pump_efficiency
)

print(f"Puissance hydraulique moyenne : "
      f"{hydraulic_power_kW.mean():.2f} kW")

print(f"Puissance pompe moyenne : "
      f"{pump_power_kW.mean():.2f} kW")
```

## 9. Récupération d'énergie sur le concentrat

Le concentrat conserve une partie de son énergie hydraulique.

On construit un modèle simplifié :

\[
P_{ERD}=P_{hyd,conc}\times\eta_{ERD}\times f_{recovery}
\]

Le terme `pressure_bar × Q_concentrate` représente ici une approximation de l'énergie hydraulique disponible dans le concentrat.

```python
concentrate_flow_m3_s = (
    concentrate_flow_m3_h * M3H_TO_M3S
)

concentrate_hydraulic_power_W = (
    pressure_pa * concentrate_flow_m3_s
)

concentrate_hydraulic_power_kW = (
    concentrate_hydraulic_power_W / 1000
)

# Fraction réellement transférable au système de pompe.
ERD_TRANSFER_FACTOR = 0.80

recovered_power_kW = (
    concentrate_hydraulic_power_kW
    * erd_efficiency
    * ERD_TRANSFER_FACTOR
)

# La récupération ne peut pas dépasser une fraction
# raisonnable de la puissance de pompe.
recovered_power_kW = np.minimum(
    recovered_power_kW,
    0.45 * pump_power_kW
)

net_power_kW = (
    pump_power_kW - recovered_power_kW
)

print(f"Puissance récupérée moyenne : "
      f"{recovered_power_kW.mean():.2f} kW")

print(f"Puissance nette moyenne : "
      f"{net_power_kW.mean():.2f} kW")
```

## 10. Consommation énergétique spécifique

\[
SEC=\frac{P_{net}}{Q_{perm}}
\]

Le résultat est en kWh/m³.

```python
SEC_physical = (
    net_power_kW / permeate_flow_m3_h
)

print(f"SEC moyenne : {SEC_physical.mean():.3f} kWh/m³")
print(f"SEC médiane : {np.median(SEC_physical):.3f} kWh/m³")
print(f"SEC min : {SEC_physical.min():.3f} kWh/m³")
print(f"SEC max : {SEC_physical.max():.3f} kWh/m³")
```

## 11. Bruit de mesure

On ajoute une variabilité d'environ 3 % à la SEC physique afin de simuler les imperfections de mesure et les variations non modélisées.

```python
noise = np.random.normal(
    loc=0,
    scale=0.03,
    size=N_SAMPLES
)

SEC_observed = np.maximum(
    SEC_physical * (1 + noise),
    0.1
)

print(f"SEC observée moyenne : "
      f"{SEC_observed.mean():.3f} kWh/m³")
```

## 12. Construction du dataset final V2

```python
df = pd.DataFrame({
    "pressure_bar": pressure_bar,
    "feed_flow_m3_h": feed_flow_m3_h,
    "salinity_g_L": salinity_g_L,
    "temperature_C": temperature_C,
    "recovery_rate_target": recovery_rate,
    "effective_recovery_rate": effective_recovery_rate,
    "pump_efficiency": pump_efficiency,
    "erd_efficiency": erd_efficiency,
    "membrane_condition": membrane_condition,

    "osmotic_pressure_feed_bar": osmotic_pressure_feed_bar,
    "osmotic_pressure_concentrate_bar": osmotic_pressure_concentrate_bar,
    "osmotic_pressure_avg_bar": osmotic_pressure_avg_bar,
    "net_membrane_pressure_bar": net_membrane_pressure_bar,

    "temperature_factor": temperature_factor,
    "membrane_factor": membrane_factor,
    "flux_permeate_L_m2_h": flux_permeate_L_m2_h,

    "permeate_flow_m3_h": permeate_flow_m3_h,
    "concentrate_flow_m3_h": concentrate_flow_m3_h,

    "hydraulic_power_kW": hydraulic_power_kW,
    "pump_power_kW": pump_power_kW,
    "recovered_power_kW": recovered_power_kW,
    "net_power_kW": net_power_kW,

    "SEC_physical_kWh_m3": SEC_physical,
    "SEC_kWh_m3": SEC_observed,
})

df.head()
```

## 13. Contrôles de cohérence

```python
print("Shape :", df.shape)
print("\nValeurs manquantes :", df.isnull().sum().sum())
print("Doublons :", df.duplicated().sum())

checks = {
    "Pression > 0":
        (df["pressure_bar"] > 0).all(),

    "Débit alimentation > 0":
        (df["feed_flow_m3_h"] > 0).all(),

    "Perméat > 0":
        (df["permeate_flow_m3_h"] > 0).all(),

    "Concentrat >= 0":
        (df["concentrate_flow_m3_h"] >= 0).all(),

    "Récupération effective <= 1":
        (df["effective_recovery_rate"] <= 1).all(),

    "Rendement pompe dans [0,1]":
        df["pump_efficiency"].between(0, 1).all(),

    "Rendement ERD dans [0,1]":
        df["erd_efficiency"].between(0, 1).all(),

    "Membrane dans [0,1]":
        df["membrane_condition"].between(0, 1).all(),

    "Pression nette > 0":
        (df["net_membrane_pressure_bar"] > 0).all(),

    "SEC > 0":
        (df["SEC_kWh_m3"] > 0).all(),
}

for name, ok in checks.items():
    print(f"{name:40s}: {'OK' if ok else 'ERREUR'}")
```

## 14. Statistiques descriptives

```python
pd.set_option("display.max_columns", None)
df.describe().T
```

## 15. Distribution de la SEC

```python
plt.figure(figsize=(9, 5))
plt.hist(df["SEC_kWh_m3"], bins=40)
plt.xlabel("SEC (kWh/m³)")
plt.ylabel("Nombre d'observations")
plt.title("Distribution de la consommation énergétique spécifique — V2")
plt.show()
```

## 16. Vérification des relations physiques

Ces graphiques servent à vérifier qualitativement que le simulateur produit des relations cohérentes.

```python
figures = [
    ("pressure_bar", "Pression (bar)", "Pression vs SEC"),
    ("salinity_g_L", "Salinité (g/L)", "Salinité vs SEC"),
    ("temperature_C", "Température (°C)", "Température vs SEC"),
    ("effective_recovery_rate", "Récupération effective", "Récupération vs SEC"),
    ("pump_efficiency", "Rendement pompe", "Rendement pompe vs SEC"),
    ("erd_efficiency", "Rendement ERD", "Rendement ERD vs SEC"),
    ("membrane_condition", "État membrane", "État membrane vs SEC"),
]

for x, xlabel, title in figures:
    plt.figure(figsize=(8, 5))
    plt.scatter(df[x], df["SEC_kWh_m3"], alpha=0.35)
    plt.xlabel(xlabel)
    plt.ylabel("SEC (kWh/m³)")
    plt.title(title)
    plt.show()
```

## 17. Corrélations avec la cible

```python
correlations = (
    df.corr(numeric_only=True)["SEC_kWh_m3"]
    .sort_values(ascending=False)
)

correlations
```

## 18. Vérification supplémentaire : bilan débit

On vérifie :

\[
Q_{feed}=Q_{perm}+Q_{conc}
\]

```python
flow_balance_error = (
    df["feed_flow_m3_h"]
    - df["permeate_flow_m3_h"]
    - df["concentrate_flow_m3_h"]
)

print(
    "Erreur maximale du bilan de débit :",
    np.abs(flow_balance_error).max(),
    "m³/h"
)
```

## 19. Sauvegarde du dataset V2

```python
df.to_csv(OUTPUT_FILE, index=False)

print("=" * 60)
print("DATASET SWRO — VERSION 2")
print("=" * 60)
print(f"Fichier      : {OUTPUT_FILE}")
print(f"Observations : {len(df)}")
print(f"Variables    : {len(df.columns)}")
print(f"SEC moyenne  : {df['SEC_kWh_m3'].mean():.3f} kWh/m³")
print(f"SEC médiane  : {df['SEC_kWh_m3'].median():.3f} kWh/m³")
print(f"SEC min      : {df['SEC_kWh_m3'].min():.3f} kWh/m³")
print(f"SEC max      : {df['SEC_kWh_m3'].max():.3f} kWh/m³")
```
