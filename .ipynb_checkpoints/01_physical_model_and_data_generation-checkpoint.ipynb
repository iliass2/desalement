# Notebook 01 — Modèle physique simplifié SWRO & génération de données

**PFE Master Science des Données**

Objectif : construire un modèle physique simplifié d'un procédé de dessalement par osmose inverse (SWRO), générer un premier dataset synthétique de 1 000 observations et vérifier sa cohérence.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

np.random.seed(42)

N_SAMPLES = 1000
OUTPUT_DIR = Path("data")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_FILE = OUTPUT_DIR / "dataset_physical_v1.csv"

BAR_TO_PA = 100000
M3H_TO_M3S = 1 / 3600

print("Configuration chargée.")
```

## 1. Hypothèses et plages de simulation

Les valeurs ci-dessous sont des **plages de simulation**. Elles ne représentent pas des mesures provenant d'une station réelle.

```python
RANGES = {
    "pressure_bar": (50, 70),
    "feed_flow_m3_h": (100, 1000),
    "salinity_g_L": (32, 42),
    "temperature_C": (15, 30),
    "recovery_rate": (0.30, 0.50),
    "pump_efficiency": (0.70, 0.90),
    "erd_efficiency": (0.70, 0.95),
    "membrane_condition": (0.70, 1.00),
}

for variable, (minimum, maximum) in RANGES.items():
    print(f"{variable:25s}: {minimum} → {maximum}")
```

```python
pressure_bar = np.random.uniform(*RANGES["pressure_bar"], N_SAMPLES)
feed_flow_m3_h = np.random.uniform(*RANGES["feed_flow_m3_h"], N_SAMPLES)
salinity_g_L = np.random.uniform(*RANGES["salinity_g_L"], N_SAMPLES)
temperature_C = np.random.uniform(*RANGES["temperature_C"], N_SAMPLES)
recovery_rate = np.random.uniform(*RANGES["recovery_rate"], N_SAMPLES)
pump_efficiency = np.random.uniform(*RANGES["pump_efficiency"], N_SAMPLES)
erd_efficiency = np.random.uniform(*RANGES["erd_efficiency"], N_SAMPLES)
membrane_condition = np.random.uniform(*RANGES["membrane_condition"], N_SAMPLES)

print("Variables d'entrée générées.")
```

## 2. Débit de perméat et de concentrat

```python
permeate_flow_m3_h = feed_flow_m3_h * recovery_rate
concentrate_flow_m3_h = feed_flow_m3_h - permeate_flow_m3_h

print(f"Débit de perméat moyen : {permeate_flow_m3_h.mean():.2f} m³/h")
print(f"Débit de concentrat moyen : {concentrate_flow_m3_h.mean():.2f} m³/h")
```

## 3. Corrections simplifiées : salinité, température et état de membrane

Ces facteurs servent à rendre le simulateur plus réaliste. Ils constituent des **hypothèses de modélisation** qui devront être discutées et justifiées dans le mémoire.

```python
salinity_reference = 35.0
salinity_factor = 1 + 0.015 * (salinity_g_L - salinity_reference)

temperature_reference = 25.0
temperature_factor = 1 + 0.008 * (temperature_reference - temperature_C)

membrane_factor = 1 + 0.30 * (1 - membrane_condition)

effective_pressure_bar = (
    pressure_bar
    * salinity_factor
    * temperature_factor
    * membrane_factor
)

print(f"Pression effective moyenne : {effective_pressure_bar.mean():.2f} bar")
```

## 4. Puissance hydraulique

Modèle simplifié :

\[
P_{hyd} = \Delta P \times Q
\]

avec la pression en Pa et le débit en m³/s.

```python
pressure_pa = effective_pressure_bar * BAR_TO_PA
feed_flow_m3_s = feed_flow_m3_h * M3H_TO_M3S

hydraulic_power_W = pressure_pa * feed_flow_m3_s
hydraulic_power_kW = hydraulic_power_W / 1000

print(f"Puissance hydraulique moyenne : {hydraulic_power_kW.mean():.2f} kW")
```

## 5. Puissance électrique de la pompe

```python
pump_power_kW = hydraulic_power_kW / pump_efficiency

print(f"Puissance pompe moyenne : {pump_power_kW.mean():.2f} kW")
```

## 6. Récupération d'énergie

On utilise ici une représentation simplifiée d'un Energy Recovery Device (ERD). Le coefficient `0.35` est un **paramètre de simulation**, pas une constante universelle.

```python
recovery_energy_kW = (
    pump_power_kW
    * (1 - recovery_rate)
    * erd_efficiency
    * 0.35
)

net_power_kW = pump_power_kW - recovery_energy_kW

print(f"Énergie récupérée moyenne : {recovery_energy_kW.mean():.2f} kW")
print(f"Puissance nette moyenne : {net_power_kW.mean():.2f} kW")
```

## 7. Consommation énergétique spécifique (SEC)

\[
SEC = \frac{P_{net}}{Q_{perméat}}
\]

Avec kW / (m³/h), on obtient directement kWh/m³.

```python
SEC_physical = net_power_kW / permeate_flow_m3_h

print(f"SEC moyenne : {SEC_physical.mean():.3f} kWh/m³")
print(f"SEC min     : {SEC_physical.min():.3f} kWh/m³")
print(f"SEC max     : {SEC_physical.max():.3f} kWh/m³")
```

## 8. Ajout d'un bruit de simulation

On ajoute environ 3 % de variabilité afin d'éviter un dataset parfaitement déterministe.

```python
noise = np.random.normal(0, 0.03, N_SAMPLES)
SEC_observed = np.maximum(SEC_physical * (1 + noise), 0.1)

print(f"SEC observée moyenne : {SEC_observed.mean():.3f} kWh/m³")
```

## 9. Construction du dataset

```python
df = pd.DataFrame({
    "pressure_bar": pressure_bar,
    "feed_flow_m3_h": feed_flow_m3_h,
    "salinity_g_L": salinity_g_L,
    "temperature_C": temperature_C,
    "recovery_rate": recovery_rate,
    "pump_efficiency": pump_efficiency,
    "erd_efficiency": erd_efficiency,
    "membrane_condition": membrane_condition,
    "effective_pressure_bar": effective_pressure_bar,
    "permeate_flow_m3_h": permeate_flow_m3_h,
    "concentrate_flow_m3_h": concentrate_flow_m3_h,
    "hydraulic_power_kW": hydraulic_power_kW,
    "pump_power_kW": pump_power_kW,
    "recovery_energy_kW": recovery_energy_kW,
    "net_power_kW": net_power_kW,
    "SEC_physical_kWh_m3": SEC_physical,
    "SEC_kWh_m3": SEC_observed,
})

df.head()
```

## 10. Contrôles de qualité

```python
print("Shape :", df.shape)
print("\nValeurs manquantes :")
print(df.isnull().sum())
print("\nDoublons :", df.duplicated().sum())

checks = {
    "Pression positive": (df["pressure_bar"] > 0).all(),
    "Débit positif": (df["feed_flow_m3_h"] > 0).all(),
    "Récupération entre 0 et 1": df["recovery_rate"].between(0, 1).all(),
    "Rendement pompe entre 0 et 1": df["pump_efficiency"].between(0, 1).all(),
    "Rendement ERD entre 0 et 1": df["erd_efficiency"].between(0, 1).all(),
    "Membrane entre 0 et 1": df["membrane_condition"].between(0, 1).all(),
    "SEC positive": (df["SEC_kWh_m3"] > 0).all(),
}

print("\nValidation physique :")
for name, ok in checks.items():
    print(f"{name:35s}: {'OK' if ok else 'ERREUR'}")
```

```python
df.describe().T
```

## 11. Visualisations de validation

```python
plt.figure(figsize=(9, 5))
plt.hist(df["SEC_kWh_m3"], bins=30)
plt.xlabel("SEC (kWh/m³)")
plt.ylabel("Nombre d'observations")
plt.title("Distribution de la SEC")
plt.show()

plt.figure(figsize=(9, 5))
plt.scatter(df["pressure_bar"], df["SEC_kWh_m3"], alpha=0.5)
plt.xlabel("Pression (bar)")
plt.ylabel("SEC (kWh/m³)")
plt.title("Pression vs SEC")
plt.show()

plt.figure(figsize=(9, 5))
plt.scatter(df["salinity_g_L"], df["SEC_kWh_m3"], alpha=0.5)
plt.xlabel("Salinité (g/L)")
plt.ylabel("SEC (kWh/m³)")
plt.title("Salinité vs SEC")
plt.show()

plt.figure(figsize=(9, 5))
plt.scatter(df["recovery_rate"] * 100, df["SEC_kWh_m3"], alpha=0.5)
plt.xlabel("Taux de récupération (%)")
plt.ylabel("SEC (kWh/m³)")
plt.title("Récupération vs SEC")
plt.show()
```

## 12. Corrélations

```python
correlation_with_target = (
    df.corr(numeric_only=True)["SEC_kWh_m3"]
    .sort_values(ascending=False)
)

correlation_with_target
```

## 13. Sauvegarde du dataset

Le fichier sera créé dans `data/dataset_physical_v1.csv`.

```python
df.to_csv(OUTPUT_FILE, index=False)

print("=" * 60)
print("DATASET SWRO — VERSION 1")
print("=" * 60)
print(f"Fichier     : {OUTPUT_FILE}")
print(f"Observations: {len(df)}")
print(f"Variables   : {len(df.columns)}")
print(f"SEC moyenne : {df['SEC_kWh_m3'].mean():.3f} kWh/m³")
print(f"SEC min     : {df['SEC_kWh_m3'].min():.3f} kWh/m³")
print(f"SEC max     : {df['SEC_kWh_m3'].max():.3f} kWh/m³")
```
