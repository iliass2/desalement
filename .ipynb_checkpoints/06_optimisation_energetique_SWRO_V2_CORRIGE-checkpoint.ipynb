Cette version :

- reconstruit les variables standardisées uniquement au moment d'appeler le MLP ;
- vérifie la convergence de l'optimiseur ;

> **Important :** les bornes physiques proposées ci-dessous sont des hypothèses de travail pour le dataset synthétique. Si des limites réelles de la station sont disponibles, elles doivent remplacer ces valeurs.

```python
# 0. Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from pathlib import Path
from scipy.optimize import differential_evolution, minimize

MODEL_DIR = Path("models")
DATA_DIR = Path("data/preprocessed")
RESULT_DIR = Path("results")
RESULT_DIR.mkdir(parents=True, exist_ok=True)

MODEL_PATH = MODEL_DIR / "best_model_optimized.pkl"

FEATURES = [
    "pressure_bar",
    "feed_flow_m3_h",
    "salinity_g_L",
    "temperature_C",
    "recovery_rate_target",
    "pump_efficiency",
    "erd_efficiency",
    "membrane_condition",
]

TARGET = "SEC_kWh_m3"

print("Configuration chargée.")
```

# 1. Charger le modèle et les données

Le MLP du Notebook 04 a été entraîné sur des variables standardisées. Nous récupérons donc le scaler sauvegardé par le pipeline précédent lorsqu'il existe.

Si le scaler n'existe pas sous ce nom, la cellule arrêtera l'exécution afin d'éviter une transformation incorrecte.

```python
# Recherche du scaler utilisé par le pipeline précédent
possible_scalers = [
    DATA_DIR / "scaler.pkl",
    MODEL_DIR / "scaler.pkl",
    MODEL_DIR / "feature_scaler.pkl",
]

SCALER_PATH = next((p for p in possible_scalers if p.exists()), None)

if not MODEL_PATH.exists():
    raise FileNotFoundError(
        f"Modèle introuvable : {MODEL_PATH}. Exécutez le Notebook 04."
    )

if SCALER_PATH is None:
    raise FileNotFoundError(
        "Scaler introuvable. Le Notebook 06 V2 refuse de reconstruire "
        "un scaler à partir du test pour éviter une incohérence."
    )

model = joblib.load(MODEL_PATH)
scaler = joblib.load(SCALER_PATH)

print("Modèle :", type(model).__name__)
print("Scaler :", type(scaler).__name__)
print("Scaler :", SCALER_PATH)
```

```python
# Données physiques : chercher le dataset original/preprocessed
possible_data = [
    DATA_DIR / "dataset.csv",
    DATA_DIR / "swro_dataset.csv",
    Path("data/dataset.csv"),
    Path("data/swro_dataset.csv"),
]

DATA_PATH = next((p for p in possible_data if p.exists()), None)

if DATA_PATH is None:
    raise FileNotFoundError(
        "Dataset physique introuvable. Placez le CSV contenant les "
        "8 variables physiques dans data/preprocessed/dataset.csv "
        "ou data/dataset.csv."
    )

df = pd.read_csv(DATA_PATH)

missing = [f for f in FEATURES if f not in df.columns]
if missing:
    raise ValueError(f"Variables physiques manquantes : {missing}")

print("Dataset :", DATA_PATH)
print("Shape :", df.shape)
display(df[FEATURES].head())
```

## 2. Fonction de prédiction sûre

Le modèle reçoit toujours les variables dans l'espace standardisé correspondant au scaler d'entraînement.

L'utilisateur travaille toutefois avec des valeurs physiques.

```python
def predict_sec_physical(X_physical):
    X_physical = pd.DataFrame(X_physical, columns=FEATURES)

    # Le scaler doit connaître exactement l'ordre des variables
    X_scaled = scaler.transform(X_physical[FEATURES])
    X_scaled = pd.DataFrame(X_scaled, columns=FEATURES)

    return np.asarray(model.predict(X_scaled), dtype=float)

# Test
X_check = df[FEATURES].head(5)
print(predict_sec_physical(X_check))
```

# 3. Vérification modèle : données physiques → MLP

Nous comparons les prédictions de cette fonction avec les prédictions du pipeline standardisé lorsqu'un jeu de test standardisé est disponible.

```python
# Si les fichiers du Notebook 04 existent, vérifier la cohérence
X_test_scaled_path = DATA_DIR / "X_test_scaled.csv"

if X_test_scaled_path.exists():
    X_test_scaled = pd.read_csv(X_test_scaled_path)[FEATURES]
    pred_direct = model.predict(X_test_scaled)

    # Transformer dans l'autre sens seulement si le scaler est compatible
    X_test_physical = pd.DataFrame(
        scaler.inverse_transform(X_test_scaled),
        columns=FEATURES
    )
    pred_physical = predict_sec_physical(X_test_physical)

    max_difference = np.max(np.abs(pred_direct - pred_physical))

    print("Différence maximale entre les deux chemins :", max_difference)

    if max_difference > 1e-6:
        raise ValueError(
            "Le scaler et le modèle ne reproduisent pas les prédictions "
            "du pipeline précédent. Vérifiez le scaler utilisé."
        )

    print("✓ Cohérence du pipeline vérifiée.")
else:
    print("X_test_scaled.csv non trouvé : vérification directe ignorée.")
```

# 4. Bornes physiques

Ces bornes sont volontairement explicites.

**À remplacer par les limites réelles de la station si elles sont disponibles.**

Pour le dataset synthétique, nous utilisons les quantiles 1 % et 99 % afin d'éviter les valeurs extrêmes.

```python
# Bornes initiales à partir des données physiques
q_bounds = pd.DataFrame({
    "min": df[FEATURES].quantile(0.01),
    "max": df[FEATURES].quantile(0.99)
})

display(q_bounds)

# Bornes opérationnelles explicites.
# Elles restent dans le domaine des données par défaut.
PHYSICAL_BOUNDS = {
    feature: (
        float(q_bounds.loc[feature, "min"]),
        float(q_bounds.loc[feature, "max"])
    )
    for feature in FEATURES
}

bounds_df = pd.DataFrame(
    PHYSICAL_BOUNDS,
    index=["min", "max"]
).T

display(bounds_df)
```

# 5. Contraintes physiques supplémentaires

Les valeurs ci-dessous servent à empêcher l'optimiseur de rechercher des solutions absurdes.

Elles sont des **hypothèses pour le système synthétique**, pas des spécifications industrielles.

```python
# Contraintes de cohérence.
# Elles peuvent être modifiées si les données réelles imposent d'autres limites.

PHYSICAL_CONSTRAINTS = {
    "recovery_rate_min": max(0.10, PHYSICAL_BOUNDS["recovery_rate_target"][0]),
    "recovery_rate_max": min(0.60, PHYSICAL_BOUNDS["recovery_rate_target"][1]),
    "pump_efficiency_min": max(0.50, PHYSICAL_BOUNDS["pump_efficiency"][0]),
    "pump_efficiency_max": min(0.95, PHYSICAL_BOUNDS["pump_efficiency"][1]),
    "erd_efficiency_min": max(0.50, PHYSICAL_BOUNDS["erd_efficiency"][0]),
    "erd_efficiency_max": min(0.95, PHYSICAL_BOUNDS["erd_efficiency"][1]),
    "temperature_min": max(5.0, PHYSICAL_BOUNDS["temperature_C"][0]),
    "temperature_max": min(40.0, PHYSICAL_BOUNDS["temperature_C"][1]),
}

print(PHYSICAL_CONSTRAINTS)
```

# 6. Objectif d'optimisation

\[
\min_x \widehat{SEC}(x)
\]

Le modèle ML prédit la SEC, mais toutes les contraintes sont appliquées dans les **unités physiques**.

```python
def constraint_penalty(row):
    penalty = 0.0

    recovery = row["recovery_rate_target"]
    pump = row["pump_efficiency"]
    erd = row["erd_efficiency"]
    temp = row["temperature_C"]

    c = PHYSICAL_CONSTRAINTS

    # Contraintes physiques explicites
    if not (c["recovery_rate_min"] <= recovery <= c["recovery_rate_max"]):
        penalty += 1000.0 * abs(recovery - np.clip(
            recovery, c["recovery_rate_min"], c["recovery_rate_max"]
        ))

    if not (c["pump_efficiency_min"] <= pump <= c["pump_efficiency_max"]):
        penalty += 1000.0 * abs(pump - np.clip(
            pump, c["pump_efficiency_min"], c["pump_efficiency_max"]
        ))

    if not (c["erd_efficiency_min"] <= erd <= c["erd_efficiency_max"]):
        penalty += 1000.0 * abs(erd - np.clip(
            erd, c["erd_efficiency_min"], c["erd_efficiency_max"]
        ))

    if not (c["temperature_min"] <= temp <= c["temperature_max"]):
        penalty += 1000.0 * abs(temp - np.clip(
            temp, c["temperature_min"], c["temperature_max"]
        ))

    return penalty


def objective_physical(x):
    row = pd.Series(x, index=FEATURES)
    sec = float(predict_sec_physical(pd.DataFrame([x]))[0])
    return sec + constraint_penalty(row)


x_reference = df[FEATURES].median().values
reference_sec = float(predict_sec_physical(pd.DataFrame([x_reference]))[0])

print(f"SEC du fonctionnement de référence : {reference_sec:.6f} kWh/m³")
```

# 7. Optimisation globale

Nous lançons plusieurs optimisations avec des graines différentes.

Cela permet de vérifier que la solution n'est pas simplement due à une seule initialisation.

```python
results_global = []

for seed in [42, 123, 2026]:
    result = differential_evolution(
        objective_physical,
        bounds=[
            PHYSICAL_BOUNDS[f]
            for f in FEATURES
        ],
        seed=seed,
        maxiter=500,
        popsize=20,
        tol=1e-8,
        atol=1e-8,
        polish=True,
        workers=1
    )

    results_global.append({
        "seed": seed,
        "success": result.success,
        "message": result.message,
        "objective": result.fun,
        "SEC": float(predict_sec_physical(pd.DataFrame([result.x]))[0]),
        "iterations": result.nit,
        "evaluations": result.nfev,
        "x": result.x
    })

global_results_df = pd.DataFrame([
    {k:v for k,v in r.items() if k != "x"}
    for r in results_global
])

display(global_results_df)
```

# 8. Vérification de convergence

Une optimisation n'est retenue comme robuste que si les différentes graines convergent vers des solutions proches et si l'algorithme signale une convergence.

```python
sec_values = np.array([r["SEC"] for r in results_global])
best_idx = int(np.argmin(sec_values))

best_global = results_global[best_idx]
x_global = best_global["x"]
global_sec = best_global["SEC"]

print("Meilleure seed :", best_global["seed"])
print("Success :", best_global["success"])
print("Message :", best_global["message"])
print(f"SEC globale : {global_sec:.6f} kWh/m³")
print(f"Dispersion des SEC entre seeds : {sec_values.std():.8f}")

if not all(r["success"] for r in results_global):
    print("AVERTISSEMENT : au moins une optimisation n'a pas signalé la convergence.")

if sec_values.std() > 1e-3:
    print("AVERTISSEMENT : les seeds donnent des solutions sensiblement différentes.")
else:
    print("✓ Les différentes seeds donnent des SEC très proches.")
```

# 9. Solution optimale dans les unités physiques

```python
optimal_solution = pd.DataFrame({
    "Feature": FEATURES,
    "Reference": x_reference,
    "Global_Optimum": x_global
})

optimal_solution["Delta"] = (
    optimal_solution["Global_Optimum"]
    - optimal_solution["Reference"]
)

display(optimal_solution)

print(f"SEC référence : {reference_sec:.6f} kWh/m³")
print(f"SEC optimum   : {global_sec:.6f} kWh/m³")

gain_abs = reference_sec - global_sec
gain_pct = gain_abs / reference_sec * 100

print(f"Gain prédit : {gain_abs:.6f} kWh/m³ ({gain_pct:.2f} %)")
```

# 10. Vérification des frontières

Nous vérifions si l'optimum est collé aux limites.

Un optimum sur une frontière n'est pas nécessairement faux, mais il doit être interprété avec prudence.

```python
boundary_rows = []

for i, feature in enumerate(FEATURES):
    low, high = PHYSICAL_BOUNDS[feature]
    value = x_global[i]
    span = high - low

    boundary_rows.append({
        "Feature": feature,
        "Value": value,
        "Min": low,
        "Max": high,
        "Distance_to_Min_%": 100*(value-low)/(span+1e-12),
        "Distance_to_Max_%": 100*(high-value)/(span+1e-12)
    })

boundary_check = pd.DataFrame(boundary_rows)
display(boundary_check)
```

# 11. Optimisation locale autour de l'optimum

Nous utilisons L-BFGS-B pour vérifier localement la solution obtenue par Differential Evolution.

```python
local_result = minimize(
    objective_physical,
    x_global,
    method="L-BFGS-B",
    bounds=[PHYSICAL_BOUNDS[f] for f in FEATURES],
    options={
        "maxiter": 2000,
        "ftol": 1e-12,
        "gtol": 1e-8
    }
)

x_local = local_result.x
local_sec = float(
    predict_sec_physical(pd.DataFrame([x_local]))[0]
)

print("Local success :", local_result.success)
print("Local message :", local_result.message)
print(f"SEC locale : {local_sec:.6f} kWh/m³")
print(f"SEC globale : {global_sec:.6f} kWh/m³")
```

# 12. Test de robustesse

Nous perturbons légèrement chaque variable autour de l'optimum.

L'objectif n'est pas de prouver la robustesse industrielle, mais de vérifier la sensibilité du résultat du modèle.

```python
rng = np.random.default_rng(42)

N_ROBUSTNESS = 1000
X_robust = np.tile(x_global, (N_ROBUSTNESS, 1))

for j, feature in enumerate(FEATURES):
    low, high = PHYSICAL_BOUNDS[feature]
    scale = 0.02 * (high - low)

    X_robust[:, j] += rng.normal(
        0,
        scale,
        N_ROBUSTNESS
    )

    X_robust[:, j] = np.clip(
        X_robust[:, j],
        low,
        high
    )

robust_predictions = predict_sec_physical(
    pd.DataFrame(X_robust, columns=FEATURES)
)

print(f"SEC moyenne : {robust_predictions.mean():.6f}")
print(f"SEC médiane : {np.median(robust_predictions):.6f}")
print(f"SEC min : {robust_predictions.min():.6f}")
print(f"SEC max : {robust_predictions.max():.6f}")
print(f"Écart-type : {robust_predictions.std():.6f}")

plt.figure(figsize=(8,4))
plt.hist(robust_predictions, bins=30)
plt.xlabel("SEC prédite (kWh/m³)")
plt.ylabel("Fréquence")
plt.title("Robustesse autour de l'optimum")
plt.tight_layout()
plt.show()
```

# 13. Recherche d'un optimum robuste

Une solution qui atteint un minimum uniquement sur une frontière peut être peu pratique.

Nous ajoutons une pénalisation modérée de l'éloignement par rapport au centre des plages admissibles.

```python
centers = np.array([
    (PHYSICAL_BOUNDS[f][0] + PHYSICAL_BOUNDS[f][1]) / 2
    for f in FEATURES
])

half_ranges = np.array([
    (PHYSICAL_BOUNDS[f][1] - PHYSICAL_BOUNDS[f][0]) / 2
    for f in FEATURES
])

def robust_objective_physical(x):
    sec = float(predict_sec_physical(
        pd.DataFrame([x], columns=FEATURES)
    )[0])

    normalized_distance = np.mean(
        ((np.asarray(x) - centers) /
         (half_ranges + 1e-12)) ** 2
    )

    # Pénalisation volontairement faible
    return sec + 0.02 * normalized_distance

robust_result = differential_evolution(
    robust_objective_physical,
    bounds=[PHYSICAL_BOUNDS[f] for f in FEATURES],
    seed=42,
    maxiter=500,
    popsize=20,
    tol=1e-8,
    polish=True
)

x_robust = robust_result.x
robust_sec = float(
    predict_sec_physical(
        pd.DataFrame([x_robust], columns=FEATURES)
    )[0]
)

print("Success :", robust_result.success)
print("Message :", robust_result.message)
print(f"SEC optimum robuste : {robust_sec:.6f} kWh/m³")
```

# 14. Comparaison des scénarios

```python
scenarios = pd.DataFrame([
    ["Reference", reference_sec],
    ["ML global optimum", global_sec],
    ["ML robust optimum", robust_sec],
    ["Local verification", local_sec],
], columns=["Scenario", "SEC_kWh_m3"])

scenarios["Gain_vs_Reference_%"] = (
    (reference_sec - scenarios["SEC_kWh_m3"])
    / reference_sec * 100
)

display(scenarios)

plt.figure(figsize=(9,4))
plt.bar(
    scenarios["Scenario"],
    scenarios["SEC_kWh_m3"]
)
plt.ylabel("SEC prédite (kWh/m³)")
plt.title("Comparaison des scénarios")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# 15. Comparaison des conditions de fonctionnement

```python
conditions = pd.DataFrame({
    "Feature": FEATURES,
    "Reference": x_reference,
    "Global_Optimum": x_global,
    "Robust_Optimum": x_robust
})

display(conditions)
```

# 16. Vérification du domaine d'apprentissage

L'optimum doit rester dans les limites réellement observées par le dataset.

C'est une protection contre une extrapolation excessive.

```python
domain_rows = []

for i, feature in enumerate(FEATURES):
    observed_min = df[feature].min()
    observed_max = df[feature].max()

    domain_rows.append({
        "Feature": feature,
        "Global_Optimum": x_global[i],
        "Observed_Min": observed_min,
        "Observed_Max": observed_max,
        "Inside_Observed_Domain": (
            observed_min <= x_global[i] <= observed_max
        )
    })

domain_check = pd.DataFrame(domain_rows)
display(domain_check)

print(
    "Optimum entièrement dans le domaine observé :",
    domain_check["Inside_Observed_Domain"].all()
)
```

# 17. Export des résultats

```python
optimal_solution.to_csv(
    RESULT_DIR / "optimization_v2_conditions.csv",
    index=False
)

boundary_check.to_csv(
    RESULT_DIR / "optimization_v2_boundary_check.csv",
    index=False
)

domain_check.to_csv(
    RESULT_DIR / "optimization_v2_domain_check.csv",
    index=False
)

conditions.to_csv(
    RESULT_DIR / "optimization_v2_scenarios_conditions.csv",
    index=False
)

scenarios.to_csv(
    RESULT_DIR / "optimization_v2_scenarios.csv",
    index=False
)

pd.DataFrame({
    "Metric": [
        "Reference_SEC",
        "Global_Optimum_SEC",
        "Robust_Optimum_SEC",
        "Local_Verification_SEC",
        "Global_Gain_percent"
    ],
    "Value": [
        reference_sec,
        global_sec,
        robust_sec,
        local_sec,
        gain_pct
    ]
}).to_csv(
    RESULT_DIR / "optimization_v2_metrics.csv",
    index=False
)

print("Résultats exportés dans :", RESULT_DIR.resolve())
```
